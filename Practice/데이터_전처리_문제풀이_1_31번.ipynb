### 2. 주요 개념 설명

#### ① 결측치·중복 데이터 처리
- `isnull()` 또는 `isna()`로 결측 여부를 확인하고, `sum()`으로 결측 개수를 계산합니다.
- `dropna()`는 기본적으로 결측값이 있는 **행**을 제거하며, `axis=1`을 지정하면 **열**을 제거합니다.
- `drop_duplicates()`는 중복 행을 제거하고, `subset`으로 비교할 열을 제한할 수 있습니다.

#### ② 이상치 탐지 및 IQR
- 제1사분위수(Q1)와 제3사분위수(Q3)의 차이인 **IQR = Q3 − Q1**을 계산합니다.
- 일반적인 후보 경계는 **Q1 − 1.5×IQR**부터 **Q3 + 1.5×IQR**까지입니다.
- `between()`으로 범위 안의 행을 선택할 수 있지만, 실제 데이터에서 이상치 삭제 여부는 업무 맥락을 고려해야 합니다.

#### ③ 범주형 데이터 인코딩
- **원-핫 인코딩**은 범주마다 0/1 더미 열을 생성하며, `pd.get_dummies()`로 구현합니다.
- **레이블 인코딩**은 범주를 정수로 변환하며, `LabelEncoder.fit_transform()`을 사용합니다. 정수 코드가 반드시 순서나 크기를 의미하지는 않습니다.
- `replace()`와 `astype(int)`로 Yes/No 또는 Y/N을 1/0으로 치환할 수 있습니다.

#### ④ 데이터 구간화와 날짜·시간 변환
- `pd.cut()`의 `bins`는 구간 경계, `labels`는 구간 이름입니다. **라벨 개수는 경계 개수보다 하나 적어야** 합니다.
- `pd.to_datetime()`으로 문자열을 날짜·시간형으로 바꾼 후 `.dt.dayofweek`와 `.dt.hour`로 요일과 시간을 추출합니다.

#### ⑤ 학습·검증 데이터 분리
- `train_test_split()`에서 `test_size=0.2`는 검증 데이터 비율 20%를 뜻합니다.
- 분류 문제에서는 `stratify=y`를 통해 학습·검증 데이터의 클래스 비율을 가급적 유지합니다.
- `random_state`는 데이터 분할의 재현성을 높입니다.

#### ⑥ 데이터 스케일링과 정보 누수 방지
- `StandardScaler`는 훈련 데이터의 평균·표준편차를 기준으로 표준화합니다.
- `MinMaxScaler`는 학습된 최솟값·최댓값을 기준으로 값을 변환합니다.
- 모델 평가에서는 **훈련 데이터에만 `fit` 또는 `fit_transform`**을 적용하고, 검증 데이터에는 **`transform`만** 적용해야 합니다.

## 08번 · 매출 IQR 이상치 제거

**유형:** 코딩

**핵심 개념:** Q1, Q3로 IQR을 계산하고 [Q1-1.5IQR, Q3+1.5IQR] 범위 안의 행만 남깁니다.

**정답:** data_clean에 정상 범위 행 저장


### 초기 데이터

이 문제를 다시 실행할 때 아래 셀부터 실행합니다.


```python
data = pd.DataFrame({'date_id':range(1,13),'revenue':[320000,340000,310000,355000,330000,345000,315000,360000,325000,1250000,338000,322000]})
```

### 정답 코드 및 실행 결과 확인


```python
q1 = data['revenue'].quantile(.25)
q3 = data['revenue'].quantile(.75)
iqr = q3-q1
lower_fence = q1-1.5*iqr
upper_fence = q3+1.5*iqr
data_clean = data.loc[data['revenue'].between(lower_fence,upper_fence)].reset_index(drop=True)
print('남은 행 수:',data_clean.shape[0])
```

**유형:** 코딩

**핵심 개념:** IQR은 Q3-Q1이며 유지 조건은 하한 이상 AND 상한 이하입니다.

**정답:** data_temp에 정상 범위 행 저장


### 초기 데이터

```python
data = pd.DataFrame({'member_id':range(1,14),'session_count':[12,15,10,14,13,11,16,9,60,12,14,10,13]})
```

### 정답 코드 및 실행 결과 확인


```python
q1 = data['session_count'].quantile(.25)
q3 = data['session_count'].quantile(.75)
iqr = q3-q1
lower_fence = q1-1.5*iqr
upper_fence = q3+1.5*iqr
data_temp = data.loc[data['session_count'].between(lower_fence,upper_fence)].reset_index(drop=True)
print('남은 행 수:',data_temp.shape[0])
```

## 14번 · 중고차 등급 LabelEncoder

**핵심 개념:** LabelEncoder는 범주를 정수로 변환하며 classes_로 코드 대응 순서를 확인할 수 있습니다.

## 18번 · 나이 구간화

**유형:** 코딩

**핵심 개념:** pd.cut()은 bins 경계와 labels를 지정하여 연속형 값을 범주화합니다. 원문 라벨 20대는 경계상 0~29세입니다.

**정답:** twenties_count에 첫 구간 수 저장


### 초기 데이터

이 문제를 다시 실행할 때 아래 셀부터 실행합니다.


```python
data = pd.DataFrame({'age':[19,23,28,34,41,45,52,58,63,25,33,47,55,21]})
```

### 정답 코드 및 실행 결과 확인


```python
data['age_group'] = pd.cut(data['age'],bins=[0,29,44,59,100],labels=['20대','30-40대','50대','60대이상'],include_lowest=True)
twenties_count = int(data['age_group'].eq('20대').sum())
print('첫 구간 수:',twenties_count)
```

## 19번 · 점수 구간화 오류 수정

**유형:** 코딩

**핵심 개념:** bins 경계가 6개이면 labels는 5개여야 합니다. 오른쪽 경계가 포함됩니다.

**정답:** a_count에 A등급 수 저장


### 초기 데이터

이 문제를 다시 실행할 때 아래 셀부터 실행합니다.


```python
data = pd.DataFrame({'score':[95,82,71,88,63,55,90,77,68,45,92,84]})
```

### 정답 코드 및 실행 결과 확인


```python
data['grade'] = pd.cut(data['score'],bins=[0,59,69,79,89,100],labels=['F','D','C','B','A'],include_lowest=True)
a_count = int(data['grade'].eq('A').sum())
print('A등급 수:',a_count)
```

## 20번 · 주문 시각에서 요일 추출

**유형:** 코딩

**정답:** monday_count에 월요일 주문 수 저장


### 초기 데이터

이 문제를 다시 실행할 때 아래 셀부터 실행합니다.


```python
data = pd.DataFrame({'order_time':['2025-01-06 10:00','2025-01-07 12:30','2025-01-06 18:00','2025-01-08 09:15','2025-01-06 20:00','2025-01-09 11:00','2025-01-06 14:20','2025-01-10 19:00','2025-01-13 10:00','2025-01-06 08:00']})
```

## 28번 · IQR의 빈칸

**유형:** 빈칸 완성

**핵심 개념:** quantile(.25)와 IQR 배수 1.5를 빈칸에 넣습니다.

**정답:** quantile / 1.5

**참고:** 28번 초기 데이터는 8번의 공개 실습 데이터를 재사용합니다.


### 초기 데이터

이 문제를 다시 실행할 때 아래 셀부터 실행합니다.


```python
data = pd.DataFrame({'date_id':range(1,13),'revenue':[320000,340000,310000,355000,330000,345000,315000,360000,325000,1250000,338000,322000]})
```

### 정답 코드 및 실행 결과 확인


```python
q1 = data['revenue'].quantile(.25)
q3 = data['revenue'].quantile(.75)
iqr = q3-q1
lower_fence = q1-1.5*iqr
upper_fence = q3+1.5*iqr
print('Q1,Q3,IQR:',q1,q3,iqr)
print('하한,상한:',lower_fence,upper_fence)
```
