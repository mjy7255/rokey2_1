## 37번 · 나이·급여 상관계수

**핵심 개념 및 정답:** corr(): Pearson 상관계수

* Pearson 상관계수 : 통계학에서 두 변수 간의 선형 상관관계의 강도와 방향을 수치로 나타낸 지표

- 범위 : [-1,1]
- 방향 :  
  *   양수 : 비례
  *   음수 : 반비례


- 강도 :
  *   0.0 ~ 0.3 : 상관관계 없음
  *   0.5 ~ 0.6 : 중간 정도 상관관계
  *   0.7 ~ 0.8 : 강한 상관관계
  *   0.9 ~ 1.0 : 매우 강한 상관관계

아래 코드를 실행하여 결과를 확인하세요.


```python
age_salary_corr = employees['age'].corr(employees['salary'])
print(age_salary_corr)
```
