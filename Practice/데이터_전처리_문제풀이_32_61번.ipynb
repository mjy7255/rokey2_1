### 2. 주요 개념 설명

#### ① 결측치 처리와 대체
- `dropna()`는 기본적으로 결측값이 있는 행을 제거합니다.
- `mean()`과 `mode().iloc[0]`으로 평균·최빈값을 구한 후 `fillna()`로 결측값을 대체합니다.

#### ② 이상치 탐지 및 강건 스케일링
- **IQR = Q3 − Q1**이며 이상치 후보 경계에 **1.5×IQR**을 자주 사용합니다.
- `RobustScaler`는 중앙값과 IQR을 사용하므로 극단값의 영향을 상대적으로 줄입니다.

#### ③ 범주형 데이터 인코딩
- `replace()`로 Yes/No 등을 1/0으로 치환하고 `astype(int)`로 자료형을 지정합니다.
- `pd.get_dummies(drop_first=True)`는 더미 변수 중 기준 범주 하나를 제외합니다.
- `LabelEncoder`의 정수 코드가 실제 크기나 순서를 의미하지 않도록 주의합니다.

#### ④ 연속형 데이터 구간화와 날짜·시간 변환
- `pd.cut()`은 `bins`와 `labels`로 연속값을 범주화합니다.
- `pd.to_datetime()`으로 날짜를 변환하고 `.dt.dayofweek`로 요일 번호를 얻습니다.

#### ⑤ 학습·검증 데이터 분리
- `test_size=0.2`는 검증 비율을 20%로 설정합니다.
- 분류 데이터는 `stratify=y`로 클래스 비율을 유지할 수 있습니다.

#### ⑥ 스케일링과 데이터 누수 방지
- 훈련 데이터에는 `fit_transform()`을, 검증 데이터에는 학습된 스케일러의 `transform()`을 사용합니다.
- 검증 데이터로 스케일러를 재학습하면 평가의 일관성이 깨집니다.

#### ⑦ 파생변수 생성과 반복 처리
- `총액 = 단가 × 수량`, `단위면적 가격 = 총가격 ÷ 면적`처럼 기존 열을 조합합니다.
- 여러 범주형 열에는 `for col in cat_cols` 반복문으로 같은 전처리를 적용할 수 있습니다.

### 32. 정수 변환 빈칸

**분류 항목:** 범주형 데이터 인코딩

**유형:** 빈칸 완성

**핵심 개념:** 문자열 Yes/No를 1/0으로 바꾼 뒤 `astype(int)`로 정수형을 지정합니다.

**정답:** `astype`

### 초기 데이터

이 문제를 다시 실행할 때 아래 셀부터 실행합니다.


```python
data=pd.DataFrame({'satisfied':['Yes','No','Yes','Yes','No','Yes','Yes','No','Yes','Yes']})
```

```python
data['satisfied']=data['satisfied'].replace({'Yes':1,'No':0}).astype(int)

print(data.head())
```

## 33. 구간화 함수 빈칸

**분류 항목:** 연속형 데이터 구간화

**유형:** 빈칸 완성

**핵심 개념:** 연속형 나이를 지정한 경계와 라벨로 분류합니다. `pd.cut()`을 사용합니다.

**정답:** `cut`

### 초기 데이터

이 문제를 다시 실행할 때 아래 셀부터 실행합니다.


```python
data=pd.DataFrame({'age':[19,23,28,34,41,45,52,58,63,25,33,47,55,21]})
```

## 34. 날짜·요일 빈칸

**분류 항목:** 날짜·시간 데이터 변환

**유형:** 빈칸 완성

**핵심 개념:** 문자열 날짜를 datetime으로 바꾸고 `.dt.dayofweek`로 월요일=0부터 일요일=6까지 추출합니다.

```python
data=pd.DataFrame({'order_time':['2025-01-06 10:00','2025-01-07 12:30','2025-01-06 18:00','2025-01-08 09:15','2025-01-06 20:00','2025-01-09 11:00','2025-01-06 14:20','2025-01-10 19:00','2025-01-13 10:00','2025-01-06 08:00']})
```

## 35. 검증 비율 빈칸

**분류 항목:** 학습·검증 데이터 분리

**핵심 개념:** 훈련:검증=8:2이면 `test_size=0.2`입니다.

### 초기 데이터

이 문제를 다시 실행할 때 아래 셀부터 실행합니다.


```python
size=[42,55,38,61,47,50,33,58,45,52,39,63,48,56,41,60,44,53,37,49]
price=[x*1000+5000 for x in size]
data=pd.DataFrame({'size':size,'price':price})
X=data.drop(columns=['price']);y=data['price']
```

## 36. 층화 기준 빈칸
